```python
# CARGAR LOS DATOS
import pandas as pd
import numpy as np

df = pd.read_excel("usoDiario.xlsx", header=17) 
df = df.dropna(how="all")

df.columns
```

```python
usoDiario = df["xi(uso diario redes sociales)"]
```

```python
df = df.dropna(subset=["xi(uso diario redes sociales)"]).head(6) #ELIMINAMOS LAS OTRAS COLUMNAS

# Aseguramos que las frecuencias absolutas sean números enteros
fi = df["fi"].astype(int).values
total_n = fi.sum() # Debería dar 200

# Definimos los intervalos y las marcas de clase (mc) para el cálculo de datos agrupados
intervalos = df["xi(uso diario redes sociales)"].values
mc = np.array([1, 3, 5, 7, 9, 11]) # Puntos medios de los intervalos (0-2, 2-4, etc.)

# ==========================================
# RECONSTRUCCIÓN DE LA TABLA EN PYTHON
# ==========================================
tabla = pd.DataFrame()
tabla["Intervalo"] = intervalos
tabla["fi"] = fi
tabla["Fi"] = tabla["fi"].cumsum()
tabla["fri"] = tabla["fi"] / total_n
tabla["Fri"] = tabla["fri"].cumsum()
tabla["pi%"] = tabla["fri"] * 100
tabla["Pi%"] = tabla["Fri"] * 100

print("--- TABLA DE FRECUENCIAS COMPLETA (Fase 4) ---")
print(tabla.to_string(index=False))
print("-" * 50)

# ==========================================
# CÁLCULOS ESTADÍSTICOS PARA DATOS AGRUPADOS
# ==========================================
# calculamos media Aritmética (Suma de mc * fi / N)
media = np.sum(mc * fi) / total_n

# calculamos varianza y Desvío Estándar para datos agrupados
varianza = np.sum(fi * (mc - media) ** 2) / total_n
desvio_estandar = np.sqrt(varianza)
coeficiente_variacion = (desvio_estandar / media) * 100

print("\n--- ESTADÍSTICAS GENERADAS EN PYTHON ---")
print(f"Media aritmética: {media:.2f} horas diarias")
print(f"Varianza: {varianza:.2f}")
print(f"Desvío estándar: {desvio_estandar:.2f} horas")
print(f"Coeficiente de variación: {coeficiente_variacion:.2f}%")
```
