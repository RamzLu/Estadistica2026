```python
# CARGAR LOS DATOS
import pandas as pd
import numpy as np

df = pd.read_excel("usoDiario.xlsx", header=17)
df = df.dropna(how="all")

df.columns
```

```python
usoDiario = df["xi(uso diario redes sociales)"]
```

```python
df = df.dropna(subset=["xi(uso diario redes sociales)"]).head(6) #ELIMINAMOS LAS OTRAS COLUMNAS

# Aseguramos que las frecuencias absolutas sean números enteros
fi = df["fi"].astype(int).values
total_n = fi.sum() # Debería dar 200

# Definimos los intervalos y las marcas de clase (mc) para el cálculo de datos agrupados
intervalos = df["xi(uso diario redes sociales)"].values
mc = np.array([1, 3, 5, 7, 9, 11]) # Puntos medios de los intervalos (0-2, 2-4, etc.) Como no sabemos exactamente los valores individuales, 
#tomas el punto medio del intervalo como representante de los valores individuales :)
#para (2-4) es (2+4)/2 = 3, y lo mismo sucesivamente


# ==========================================
# RECONSTRUCCIÓN DE LA TABLA EN PYTHON
# ==========================================
tabla = pd.DataFrame() #cremos un DataFrame nuevo y le vamos agregando columnas y sus valores

tabla["Intervalo"] = intervalos
tabla["fi"] = fi #Frecuencia absoluta, que nos dice cuántas veces ocurre
tabla["Fi"] = tabla["fi"].cumsum() #la función .cumsum() hace la suma acumulada (suma las frecuencias absolutas anteriores)
tabla["fri"] = tabla["fi"] / total_n #tomamos cada frecuencia absoluta (fi) y la dividimos por el total (total_n). Eso nos dice qué proporción de la muestra pertenece a esa clase (ej. 0.15)
tabla["Fri"] = tabla["fri"].cumsum() # es lo mismo que la frecuencia absoluta acumulada, pero con las relativa
tabla["pi%"] = tabla["fri"] * 100 #s la frecuencia relativa multiplicada por 100 para tenerlo en formato porcentaje (siguiendo el ej. anterior, 15%)
tabla["Pi%"] = tabla["pi%"].cumsum() #es la frecuencia relativa acumulada multiplicada por 100, o así también la suma consecutiva de pi%

print("--- TABLA DE FRECUENCIAS COMPLETA (Fase 4) ---")
print(tabla.to_string(index=False))
print("-" * 50)

# ==========================================
# CÁLCULOS ESTADÍSTICOS PARA DATOS AGRUPADOS
# ==========================================

# calculamos media Aritmética (Suma de mc * fi / N)
media = np.sum(mc * fi) / total_n #para calcular el promedio con datos agrupados, multiplicas cada marca de clase (mc) 
#por su frecuencia (fi), sumas todos esos resultados, y luego lo divides por el total de la muestra (total_n)

# calculamos varianza y Desvío Estándar para datos agrupados

varianza = np.sum(fi * (mc - media) ** 2) / total_n #mide qué tan "esparcidos" están los datos alrededor de la media. Para cada intervalo, se
#toma la marca de clase, se resta la media, se eleva el resultado al cuadrado, lo multiplicas por la frecuencia de ese intervalo, sumas todo y divides por el total.

desvio_estandar = np.sqrt(varianza)#es simplemente la raíz cuadrada (np.sqrt) de la varianza, es un poquito mas facil de entender porq syu unidad de medida es en horas.

coeficiente_variacion = (desvio_estandar / media) * 100#mide la relación entre el desvío estándar y la media como un porcentaje, 
#sirve para comparar la dispersión de dos conjuntos de datos diferentes


print("\n--- ESTADÍSTICAS GENERADAS EN PYTHON ---")
print(f"Media aritmética (medida pertinente de comparación): {media:.2f} horas diarias")
print(f"Varianza: {varianza:.2f}")
print(f"Desvío estándar: {desvio_estandar:.2f} horas")
print(f"Coeficiente de variación: {coeficiente_variacion:.2f}%")
```
